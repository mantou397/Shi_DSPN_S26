## 3. Principal component analysis (3 points)

Let's see how many principal components explain at *at least* 95% of the data.

(a) Create a new object called `fs_d.pca` using the `princomp` function (do not forget to scale the data).


```r
fs_d <- d1 %>% select(starts_with("FS_"))
fs_d.pca <- princomp(fs_d, cor = TRUE)
```

(b) Calculate the cumulative variance explained (not unique variance explained as in the tutorial) across the principal components and plot the results using `ggplot`.

```r
variance_explained <- summary(fs_d.pca)$sdev^2 / sum(summary(fs_d.pca)$sdev^2)
cumulative_variance <- cumsum(variance_explained)

cum_var_df <- data.frame(
  PC = 1:length(cumulative_variance),
  Cumulative_Variance = cumulative_variance
)

ggplot(cum_var_df, aes(x = PC, y = Cumulative_Variance)) +
  geom_line() +
  geom_point() +
  geom_hline(yintercept = 0.95, linetype = "dashed", color = "red") +
  labs(title = "Cumulative Variance Explained by Principal Components",
       x = "Principal Component",
       y = "Cumulative Proportion of Variance Explained") +
  theme_minimal()
```

(c) Determine exactly how many principal components explain at least 95% of the variance.

Hint: Look up the `which` function.

```r
n_components <- which(cumulative_variance >= 0.95)[1]
n_components
```

What does this tell you about the underlying dimensionality of the brain volume measures?

> The results tells us that the brain volume data has moderately high intrinsic dimensionality. Since out of the origional freesurfer volume measures, 27 principal components are still needed to capture 95% of variance, so the data cannot easily be reduced just to a few dimensions. Also, from the figure, we can see that the first few components capture a disproportionate share of variance, where the curve rises steeply at first, and then start to reach the ceiling. This steep initial rise may indicate the strong shared variance we saw in the heatmap. Also, the curve flattens out gradually rrather than sharply. This could mean that there is no clear cutpoint where a small number of components captures nearly everything. So beyond global brain size factor, there are still many additional independent sources of variation that may not be explained by a single general size factor.

## 4. Associating with Flanker task peformance (4 points)

Now apply PCR to the `d1` object you created at the beginning (which includes the Flanker task scores) to find how freesurfer volumes predict Flanker task performance. Set the random seed to "2". Use cross-validation as the validation type and don't forget to scale your data. Show the `summary` of the model fit.

Hint: If you receive an error applying the "scale=TRUE" flag, then you likely still have columns of all zeros somewhere in your data table.

```r
set.seed(2)
pcr_model <- pcr(Flanker_Unadj ~ ., data = d1, scale = TRUE, validation = "CV")
summary(pcr_model)
```

(b) Use the `validationplot` function to evaluate the bias-variance tradeoff using the cross-validated mean squared-error for each component.

```r
validationplot(pcr_model, val.type = "MSEP")
```

(c) Extract the MSEP values of the cross validated fit ("CV" not "adj_CV") from the model object using the `MSEP` funtion. Create a new array of these values. Use the `drop` function to remove the singleton dimension (the original array is 2x1x53 and we want a 2x53 object). Find the minimum value of the first row (the "CV").

```r
msep_vals <- MSEP(pcr_model)$val
msep_vals <- drop(msep_vals)  

min_msep <- min(msep_vals[1, ])
min_msep
```

What what does this plot tell you about how many components best explain variance in Flanker task performance?

> The validation plot shows that the minimum cross-validated MSEP (~97.80) occurs at approximately 7-10 components. After that point, MSEP begins to rise, which could mean that adding more components can lead to overfitting and the model starts fitting noise rather than meaningful signal. Moreover, with too few components (like around 0 components), the model underfits with a very high MSEP. The error is minimized at around 7-10 components and beyond that, prediction error steadily climbes back as the model becomes overfit. So roughly 7-10 principal components best predict Flanker task performance from brain volume measures.
